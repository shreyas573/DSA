##  moving zeros at the end 

```python
def zeros_end(array): 
    non_zero=[x for x in array if x!=0]
    zeros=[x for x in array if x==0]
    return non_zero+zeros
input_array=[0,23,9,0]
zeros_end(input_array)
```

## count the no of vowels 

```python
letter="shreyas"
vowels="aeiou"
count=0
v=[]
for ch in letter:
    if ch in vowels:
        count+=1
        v.append(ch)
print("vowels preseent in the letter:",count)
print("vowels are :",v)
```

## prime 

```python
def prime( number): 
    num=int(number**0.5)+1
    for i in range(2,num):
        if number%i==0:
            return False 
        return True
inp=int(input())
prime(inp)
```
