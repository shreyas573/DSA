### Pair sum


```python
number=[2,3,4,5,6]
target=7
seen=set()
for arr in number: 
    complement=target-arr
    if complement in seen:
        print(complement,arr)
    seen.add(arr)
```

```python
number=[6,5,7,3,8,9]
target=13
for i in range(len(number)):
    for j  in range(i+1,len(number)):
        if number[i]+number[j]==target:
            print(i,j)
```

###  move zeros at the end 

```python
number=[1,2,3,0,5,0,3,7]

zeros=[x for x in number if x==0]
non_zeros=[x for x in number if x!=0]
print(non_zeros+zeros)
```

### Reverse string 

```python
def reverse(s:str)->str:
    text=""
    for char in range(len(s)-1,-1,-1):
        text+=s[char]
    return  text
name="shreyas"
reverse(name)
```

### Longest substring   without ends

```python
s = "abcabcbb"

left = 0
max_length = 0
char_set = set()

for right in range(len(s)):

    while s[right] in char_set:
        char_set.remove(s[left])
        left += 1

    char_set.add(s[right])

    max_length = max(max_length, right - left + 1)

print(max_length)
```
