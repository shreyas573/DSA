## divisor problem

```python
n=6
count=0
for i in range(1,int(n**0.5)+1):
    if n%i==0:
        if i==n//i:
            count+=1
        else: 
            count+=2
            
print(count)
        
```

## perfect number

```python
n = 28
total = 0

if n > 1:
    total = 1

    for i in range(2, int(n ** 0.5) + 1):

        if n % i == 0:

            if i == n // i:
                total += i

            else:
                total += i + (n // i)

if total == n:
    print("Perfect Number")
else:
    print("Not a Perfect Number")
```

## prime number

```python
n=8

is_prime=True
if n<2:
    print("not prime")
for i in (2,int(n**0.5)+1):
    if(n%i==0):
        is_prime=False
        break
    else:
        is_prime=True
        
if(is_prime):
    print("true")
else:
    print("false")
```

## Nth prime number

```python
n = 10
num = 2
count = 0

while count < n:

    is_prime = True

    for i in range(2, int(num ** 0.5) + 1):

        if num % i == 0:
            is_prime = False
            break

    if is_prime:
        count += 1

    num += 1

print(num - 1)
```
